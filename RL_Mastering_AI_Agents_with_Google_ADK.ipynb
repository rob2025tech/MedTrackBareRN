## 🧠 Part 2: Basic Concepts of Agents
### **The Agentic Loop: 5 Steps to Autonomy**

<br>

An agent doesn't just guess; it follows a structured **"Think → Act → Observe"** reasoning path.

<br>

| Step | Phase | Action Description | Example (Help me return order #123) |
| :--- | :--- | :--- | :--- |
| **1** | **Receive Task** | Understand intent & extract entities. | *"User wants a return. Order ID detected: #123."* |
| **2** | **Scan Environment** | Check available tools & context. | *"I have `get_order_details` and `process_refund` tools."* |
| **3** | **Plan** | Break the goal into logical sub-tasks. | *"1. Verify order status. 2. Check return policy."* |
| **4** | **Act** | Execute the specific tool call. | `call get_order_details(order_id="123")` |
| **5** | **Observe and Iterate** | Evaluate the result & decide next step. | *"Order exists. Item is eligible. Triggering refund..."* |

<br>


> This **'Think, Act, Observe'** cycle continues — managed by the **Orchestration Layer**, reasoned by the **Model**, and executed by the **Tools** until the initial **Mission** is achieved.

## 🛠️ Section 3: Basic Components in Building an Agent
### **The LLM Agent**

<br>

An **LLM Agent** is the most common type of agent in ADK. It relies on the model's ability to reason which tool to use.

<br>

**Core Implementation:**
```python
from google.adk.agents import Agent
from google.adk.tools import google_search


root_agent = Agent(
    name="basic_search_agent",
    model="gemini-3-pro-preview",
    description="Agent to answer questions using Google Search.",
    instruction="""
    You are a helpful assistant with access to Google Search.
    
    If the user asks a question that requires current information or facts, use the 'google_search' tool.
    Always cite your sources implicitly by providing the answer clearly based on the search results.
    """,
    # google_search is a pre-built tool which allows the agent to perform Google searches.
    tools=[google_search]
)

```

## 🛠️ Section 3: Basic Components in Building an Agent
### **Agentic Workflows**

<br>

ADK allows you to define how an agent moves through a task. Choosing the right **Workflow** ensures your agent is efficient, reliable, and capable of handling errors.

<br>



<br>

##### **1. Sequential Workflow (The Linear Path)**
* **The Flow:** Step-by-step order (1 → 2 → 3).
* *Use Case:* When Step B depends on the output of Step A.
* *Example:* **Summarize** a meeting transcript → **Email** the team.

<br>

##### **2. Parallel Workflow (The Multi-Tasker)**
* **The Flow:** Multiple tools or tasks triggered simultaneously.
* *Use Case:* To reduce latency when tasks are independent.
* *Example:* **Price Comparison** — checking three different vendors at the same time.

<br>

##### **3. Loop / Cycle (The Iterative Solver)**
* **The Flow:** Repeating a step until a "Success" condition is met or a limit is reached.
* *Use Case:* For self-correction, debugging code, or refining a creative draft.
* *Example:* **Write Code** → **Run Test** → **Fix Error** (Repeat until tests pass).

<br>


> ADK's `WorkflowAgent` gives you the power to mix these patterns, creating robust pipelines that don't just "guess," but follow your logic.

## 🚀 Part 5: Beyond the Basics
### **Advanced Capabilities from "Advent of Agents"**

<br>

Once you master the basic "Think-Act-Observe" loop, the **Google ADK** offers powerful enterprise-grade features that solve real-world UX and performance challenges.

<br>

#### **1. Context Caching & Compaction (Day 10)**
* **What is it:** Instead of re-sending a 100-page PDF with every message, ADK **caches and compressing** the data on the server via *compaction_interval / overlap_size*.
* **Benefit:** Reduces latency by up to 80% and significantly cuts token costs for long conversations.

<br>

#### **2. Session Rewind: "Time Travel" (Day 09)**
* **What is it:** Built-in ability to "undo," edit previous messages, or branch out into a new conversation path via *rewind_async*.
* **Benefit:** Users can correct their mistakes without the Agent getting confused by old, incorrect context.

<br>

#### **3. Multimodal Streaming (Day 12)**
* **What is it:** The agent doesn't just send text; it can stream audio, images, and tool outputs simultaneously.
* **Benefit:** Creates a highly responsive, "Live" feeling interaction (like Gemini Live).
